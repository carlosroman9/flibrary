## 📥 Fallback download (if the UI download buttons are blocked)
Run this after clicking **Clean**. It downloads the files from the most recent run.

```python
import json, os
from google.colab import files
from marc_cleaner.ui import LAST_RUN_FILE

if os.path.exists(LAST_RUN_FILE):
    with open(LAST_RUN_FILE) as fh:
        last = json.load(fh)
    for path in last.values():
        if path and os.path.exists(path):
            print("Downloading:", os.path.basename(path))
            files.download(path)
else:
    print("No recent output found. Click **Clean** in the UI first.")
```
